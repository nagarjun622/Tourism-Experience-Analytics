# EDA and Visualization

**Objective:** Explore tourism trends, demand, ratings, and visit modes.

The code is split into visible, explainable steps for a demonstration video.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RAW = ROOT / 'data' / 'raw' / 'Tourism Dataset'
PROCESSED = ROOT / 'data' / 'processed'
RANDOM_STATE = 42
pd.set_option('display.max_columns', 50)
```

```python
master = pd.read_csv(PROCESSED/'master_dataset.csv')
master.shape, master[['Rating','VisitMode']].describe(include='all')
```

```python
fig, axes = plt.subplots(1,3,figsize=(18,4))
master.VisitYear.value_counts().sort_index().plot.bar(ax=axes[0], title='Visits by year')
master.VisitMode.value_counts().plot.bar(ax=axes[1], title='Visit mode distribution')
master.Rating.value_counts().sort_index().plot.bar(ax=axes[2], title='Rating distribution')
plt.tight_layout(); plt.show()
```

```python
attraction_summary = master.groupby('Attraction').agg(visits=('TransactionId','size'), average_rating=('Rating','mean')).sort_values('visits', ascending=False)
attraction_summary.head(15)
```
