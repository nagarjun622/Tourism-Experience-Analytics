# Data Cleaning and Preprocessing

**Objective:** Apply justified cleaning while preserving raw workbooks.

The code is split into visible, explainable steps for a demonstration video.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RAW = ROOT / 'data' / 'raw' / 'Tourism Dataset'
PROCESSED = ROOT / 'data' / 'processed'
RANDOM_STATE = 42
pd.set_option('display.max_columns', 50)
```

```python
raw = pd.read_excel(RAW/'Transaction.xlsx')
raw.head()
```

```python
# Only exact duplicate records are removed. This action is reported.
clean = raw.drop_duplicates().copy()
cleaning_log = pd.DataFrame([{'before_rows': len(raw), 'after_rows': len(clean), 'rows_affected': len(raw)-len(clean), 'reason': 'Removed exact duplicates only'}])
cleaning_log
```

```python
pd.Series({'invalid_rating_rows': ((clean.Rating < 1) | (clean.Rating > 5)).sum(), 'invalid_month_rows': ((clean.VisitMonth < 1) | (clean.VisitMonth > 12)).sum(), 'missing_rating_rows': clean.Rating.isna().sum()})
```

```python
# Processed files are separate from the raw source files.
clean.to_csv(PROCESSED/'transaction_clean.csv', index=False)
print('Saved transaction_clean.csv')
```
